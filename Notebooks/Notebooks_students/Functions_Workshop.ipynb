#### Functions

A function is a named chunk of code that does one job. You can give it inputs (called parameters), it runs the steps inside, and it can give back a result. Functions help break a program into smaller bits so the code is easier to reuse and understand. Some functions don't return anything and just do something (like print), and some can even take other functions as inputs.
