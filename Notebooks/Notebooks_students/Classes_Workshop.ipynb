### Coding the shape class
Writing the code for the `Shape` class example could look like this:


```python
class Shape:
    def __init__(self, form, size, color):
        self.form = form        # "circle" or "square"
        self.size = size        # integer
        self.color = color      # "blue", "red" or "yellow"
```

The most important part to take from it is the `__init__` function. This function initiates the shape, and is essentially what puts the form, size and color in the box. Type in some combinations in `shape1` and `shape2` below, and run the code to see what happens.

```python
shape1 = Shape("square", 3, "blue") # Insert your combinations here
shape2 = Shape("circle", 5, "yellow") # Insert your combinations here
print(shape1)
print(shape2)
```

As you can see this doesn't draw the shape. To draw it we need to add a `draw` function. Below is an example of what it could look like, it is not necessary to understand the details of how it works, as this is just an example. The `draw` function takes two arguments, `self` and `filled`. In a class, a function should always take the `self` argument, if attributes are needed in the function. Run the code block before you continue.

```python
class Shape:
    def __init__(self, form, size, color):
        self.form = form    
        self.size = size        
        self.color = color      

    def draw(self, filled): # Filled/blank
        import matplotlib.pyplot as plt
        from matplotlib.patches import Circle, Rectangle
        fig, ax = plt.subplots()

        if self.form == "circle":
            ax.add_patch(Circle((0,0), self.size, facecolor = self.color if filled else "none", edgecolor = self.color))
            ax.set_xlim(-self.size - 1, self.size + 1)
            ax.set_ylim(-self.size - 1, self.size + 1)
            ax.set_aspect("equal")
            ax.grid(alpha = 0.3)

        elif self.form == "square":
            ax.add_patch(Rectangle((0,0),self.size,self.size, facecolor = self.color if filled else "none", edgecolor = self.color))
            ax.set_xlim(-1, self.size + 1)
            ax.set_ylim(-1, self.size + 1)
            ax.set_aspect("equal")
            ax.grid(alpha = 0.3)
        plt.show()
```

The Shape class now has a `draw` method. A method is called by adding a period after the object, and writing the name of the method, as seen below. Type in your combinations from before again, and run the code block to see what happens. Notice, that the `draw` function only takes the argument `filled`, as the `self` argument is integrated by definition. 

```python
shape1 = Shape("square", 3, "blue") # Insert your combinations here
shape2 = Shape("circle", 5, "yellow") # Insert your combinations here
shape1.draw(filled=True)
shape2.draw(filled=False)
```

Usually an imaginary number is written 
$$z=a+ib$$
where $a$ is the value of the real part and $b$ is the value for the imaginary part. An imaginary number class would thus need to consist of the value for $a$ and $b$. If for example we had the imaginary number $z=2+3i$ we would need to remember that $a=2$ and $b=3$. Writing this imaginary number in python could look like the code block below. Try to run the code blocks below, and see what they print.

```python
class ImagNum:
    def __init__(self):
        self.a = 2
        self.b = 3
```

```python
z = ImagNum()
print(z)
```

```python
print(z.a)
```

```python
print(z.b)
```

Notice, how we acces the attributes of the imaginary number. We write the name of the object and add a period followed by the attribute we want to look at, just as calling a method.
